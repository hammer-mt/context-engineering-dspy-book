## Without DSPy: a raw prompt string

Traditional prompting means brittle string replacement: you write the template, then `.format(...)` the variables in by hand.

```python
prompt = (
    "You’re an expert in {topic}. Given context: {context}, "
    "answer: {question}"
)

formatted_prompt = prompt.format(
    topic="Physics",
    context="The speed of light is 300,000 km/s.",
    question="What is the speed of light?",
)

print(formatted_prompt)
```

## With DSPy: a signature

In DSPy you declare the inputs and outputs as a class, and DSPy handles building the prompt for you.

```python
import dspy
class QA(dspy.Signature):
        """Answer the question as an expert in the topic."""
        topic: str = dspy.InputField()
        context: str = dspy.InputField()
        question: str = dspy.InputField()
        answer: str = dspy.OutputField()
```

## Run the signature

> **Book vs. notebook:** The listing on page 6 stops at the signature definition. The cells below go one step further: they run the signature with `dspy.Predict` and print the prompt DSPy builds from it, which Chapter 2 examines in detail.

```python
qa = dspy.Predict(QA)

result = qa(
    topic="Physics",
    context="The speed of light is 300,000 km/s.",
    question="What is the speed of light?",
)

print(result.answer)
```
